```python
import pandas as pd

df = pd.read_parquet('dataset/valid.all.parquet')
print(f"原始数据集行数: {len(df)}")

# 创建一个字典来存储每个数据源已经见过的索引
seen_indices = {
    'aime': set(), 
    'aime25': set(), 
    'amc': set(), 
    'math': set(), 
    'minerva': set(), 
    'olympiad_bench': set()
}

eval_data = []
duplicates_count = 0

for i in range(len(df)):
    row = df.iloc[i]
    data_source = row.get('data_source')
    
    # 获取extra_info.index，处理可能的缺失或嵌套结构
    try:
        if isinstance(row.get('extra_info'), dict):
            index = row['extra_info'].get('index')
        else:
            index = None
    except:
        index = None
    
    # 创建标识符
    identifier = (data_source, index)
    
    # 如果数据源有效且索引不为None
    if data_source in seen_indices and index is not None:
        # 检查是否已经见过这个索引
        if index in seen_indices[data_source]:
            duplicates_count += 1
            continue  # 跳过这一行
        else:
            # 记录这个索引
            seen_indices[data_source].add(index)
    
    # 添加到结果列表
    eval_data.append(row)

print(f"找到并跳过了 {duplicates_count} 个重复项")
print(f"去重后数据行数: {len(eval_data)}")

# 每个数据源去重后的项目数
for source, indices in seen_indices.items():
    if indices:
        print(f"{source}: {len(indices)} 个唯一项")

# 转换为DataFrame并保存
eval_df = pd.DataFrame(eval_data)
eval_df.to_parquet("dataset/eval.passn.parquet")
print("已保存去重后的数据到 dataset/eval.passn.parquet")
```
